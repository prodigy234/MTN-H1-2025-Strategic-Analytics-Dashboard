```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Example dataset
data = {
    "Product": ["A", "B", "C", "D", "E"],
    "Sales": [120, 340, 230, 150, 400]
}

df = pd.DataFrame(data)

# Bar chart
plt.figure(figsize=(6,4))
sns.barplot(x="Product", y="Sales", data=df)
plt.title("Sales by Product")
plt.show()
```
